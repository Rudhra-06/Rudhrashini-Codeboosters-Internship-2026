## End-to-End AI Data Systems — Architecture

An end-to-end AI Data System is a complete working product that takes raw data in and delivers intelligent answers out. It has four layers working together.

---

### Layer 1: Data Ingestion (Green)

* **Tools:** Pandas, APIs, CSV, JSON
* Collect and load raw data from all sources into the system

### Layer 2: Data Processing (Blue)

* **Tools:** ETL, Cleaning, SQLite
* Clean, transform, and structure data so it is usable by AI

### Layer 3: AI Reasoning (Purple)

* **Tools:** Groq API, RAG, ChromaDB
* Apply LLMs and embeddings to understand and reason about data

### Layer 4: Delivery + Interface (Red)

* **Tools:** Python Output, Chatbot
* Present intelligent answers to end users in a useful format

```python
conn=sqlite3.connect(':memory:')

student_df.to_sql('students', conn, if_exists='replace', index=False)

print("SQL database created.")
print(f"Table 'students' loaded with {len(student_df)} rows.")
```

```python
query1="""
SELECT
  branch,
  COUNT(*) AS total_students,
  ROUND(AVG(gpa),2) AS avg_gpa,
  ROUND(AVG(attendance_pct),1) AS avg_attendance
FROM students
GROUP BY branch
ORDER BY avg_gpa DESC"""

branch_analysis=pd.read_sql_query(query1, conn)
print("=== Branch-Wise Performance Analysis ===")
print(branch_analysis.to_string(index=False))
```

```python
query2="""
SELECT name, branch, gpa, attendance_pct, passed
FROM students
ORDER BY gpa DESC
LIMIT 5"""
top_students=pd.read_sql_query(query2, conn)
print("=== Top 5 Students ===")
print(top_students.to_string(index=False))
```

```python
query3="""
SELECT
  passed,
  COUNT(*) AS student_count,
  ROUND(AVG(gpa),2) AS avg_gpa
FROM students
GROUP BY passed"""

pass_fail=pd.read_sql(query3, conn)
print("=== Pass/Fail Statistics ===")
print(pass_fail.to_string(index=False))

total = len(student_df)
passed = len(student_df[student_df['passed']=='Yes'])
pass_rate = round((passed / total) * 100 , 1)
print(f"\nOverall Pass Rate: {pass_rate}% ({passed}/{total} students)")
```

Insights -> meaningful finding or understanding

In data engineering, an insight is a meaningful finding or understanding derived from data that helps people make decisions, identify problems, or discover opportunities.

Think of it like this:

Data = raw facts

Information = organized data

Insight = the "so what?" conclusion you can act on

```python
branch_summary=""
for _, row in branch_analysis.iterrows():
  branch_summary+=f" - {row['branch']} : {row[...]} students, Avg GPA {row['avg_gpa']}, Avg Attendance {row['avg_attendance']}\n"

top_summary=""
for _, row in top_students.iterrows():
  top_summary+=f" - {row['name']} in {row['branch']} with GPA {row['gpa']}\n"

data_summary=f"""
STUDENT PERFORMANCE DATA SUMMARY
Total Students : {total}
Overall Pass Rate : {pass_rate}%
Average GPA across all students : {round(student_df['gpa'].mean(), 2)}

Performance by branch:
{branch_summary}
Top 5 stuents by gpa:
{top_summary}"""

print("Data summary prepared")
print(data_summary)
```

```python
system_prompt="""
You are an expert academic data analyst working for an engineering college.
You recieve student performance summaries and provide clear, actionable insights.
Always base your analysis strictly to the data provided.
If you are uncertain about something, clearly say so.
Format your response with numbered points for clarity."""

user_message=f"""
Here is the student performance data for this semester:

{data_summary}

Please provide :
1. Three key insights from this data
2. Which branch needs the most improvement
3. One recommedation for the college principal"""

response=client.chat.completions.create(
    model=MODEL,
    max_tokens=200,
    messages=[
        {"role":"system", "content":system_prompt},
        {"role":"user", "content":user_message}
    ]
)

ai_analysis=response.choices[0].message.content

print("="*60)
print("AI-GENERATED DATA ANALYSIS")
print("="*60)
print(ai_analysis)
```
